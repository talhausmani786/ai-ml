```python
import pandas as pd
import numpy as np
```

```python
performance_scores = np.random.normal(loc = 75 , scale = 10 , size  = 100)
```

```python
performance_scores[10] = 150 #  high outlier
performance_scores[25] = -20 #  low outlier
performance_scores[50] = 5
```

```python
employee_df = pd.DataFrame({
'EmployeeID': range(1, 101),
'PerformanceScore': performance_scores.round(0)
})
```

```python
employee_df.head()
```

```python
Q1 = employee_df.quantile(.25)
Q1
```

```python
Q3 = employee_df.quantile(.75)
```

```python
IQR  = Q3 - Q1
```

```python
upper_bound = Q3 + 1.5*IQR
```

```python
upper_bound['PerformanceScore']
```

```python
lower_bound = Q1 - 1.5*IQR
```

```python
employee_df[(employee_df['PerformanceScore'] > upper_bound['PerformanceScore']) | (employee_df['PerformanceScore']<lower_bound['PerformanceScore'])]
```

```python
employee_df['PerformanceScore'][(employee_df['PerformanceScore']> upper_bound['PerformanceScore'])]
```

```python
employee_df.loc[(employee_df['PerformanceScore']> upper_bound['PerformanceScore']) , 'PerformanceScore'] = upper_bound['PerformanceScore']
```

```python
employee_df.loc[(employee_df['PerformanceScore']< lower_bound['PerformanceScore']) , 'PerformanceScore'] = lower_bound['PerformanceScore']
```
