```python
import pandas as pd
```

```python
movie_info = pd.read_csv(
    'u.item',
    sep='|',
    encoding='latin-1',
    header=None , names=['movie_id', 'movie_title']
)

movie_info.head()
```

```python
ratings = pd.read_csv('u.data' , sep = '\t' ,     names=['user_id', 'item_id', 'rating', 'timestamp'] )
```

```python
ratings.head()
```

```python
ratings['item_id']
```

```python
# Step 3: Calculate average rating and number of ratings per movie
movie_agg_df = ratings.groupby('item_id').agg(
    average_rating=('rating', 'mean'),
    number_of_ratings=('rating', 'count')
).reset_index()

# Step 4: Merge aggregated ratings with movie information
merged_df = pd.merge(
    movie_info,
    movie_agg_df,
    left_on='movie_id',
    right_on='item_id',
    how='inner'
)

# Step 5: Create the required DataFrame
movie_stats_df = merged_df[
    ['movie_title', 'average_rating', 'number_of_ratings']
]

# Step 6: Sort by number of ratings in descending order
movie_stats_df = movie_stats_df.sort_values(
    by='number_of_ratings',
    ascending=False
)

# Step 7: Display the top 10 movies
print(movie_stats_df.head(10).to_string(index=False))
```
